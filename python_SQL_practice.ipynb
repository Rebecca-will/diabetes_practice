# Diabetes Risk Analytics Using SQL and Python
Objective
This project uses a synthetic diabetes prediction dataset to examine data quality, population characteristics, diabetes prevalence, and relationships between selected clinical indicators and diabetes status.

The project demonstrates:

- SQL data profiling and aggregation
- Python exploratory data analysis
- Data-quality assessment


This analysis is educational and is not intended for clinical decision-making.

```python
%sql
SELECT DISTINCT smoking_history
FROM workspace.default.diabetes_prediction_dataset;
```

```python
%sql
SELECT smoking_history, COUNT(1) AS amount
FROM workspace.default.diabetes_prediction_dataset
GROUP BY smoking_history
ORDER BY amount DESC;
```

```python
%sql
SELECT diabetes, COUNT(1) as amounts
FROM workspace.default.diabetes_prediction_dataset
GROUP BY diabetes
ORDER BY amounts DESC;
```

This is a very unbalanced dataset. 
If this was used for predicition the model could achieve a very high accuracy by just labelling all the values as 0, non diabetics. 

It would be important to evaluate the recall of a model used to understand how successful the model was at identifying those with the disease


```python
%sql
SELECT MIN(bmi) AS min_bmi, MAX(bmi) AS max_bmi, AVG(bmi) AS AVG_BMI
from workspace.default.diabetes_prediction_dataset;
```

Looking at these values the max is very different from the min and the avg. This would be something to look at during EDA, to understand if the data may be right skewed.

I would also check if the max value could be considered an outlier. If possible I would verify the value for accuracy with the source as well.

```python
%sql
SELECT diabetes, ROUND(AVG(bmi),2) AS avg_bmi
FROM workspace.default.diabetes_prediction_dataset
GROUP BY diabetes
```

```python
%sql
SELECT diabetes, ROUND(AVG(age),2) AS avg_age, ROUND(AVG(BMI),2) as AVG_BMI, ROUND(avg(HBa1c_LEVEL),2) AS avg_HBa1c, ROUND(AVG(blood_glucose_level),2) AS avg_BG
FROM workspace.default.diabetes_prediction_dataset
GROUP BY diabetes
 
```

```python
%sql
SELECT hypertension, SUM(CASE WHEN diabetes =1 THEN 1 else 0 END) AS diabetes_cases
FROM workspace.default.diabetes_prediction_dataset
GROUP BY hypertension;
```
